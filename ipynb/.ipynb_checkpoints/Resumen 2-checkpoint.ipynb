## Introduccion

- El capitulo 2 se centra en los problemas de inferencia o predicciones dentro de la clase o el aprendizaje supervisado.
- Dentro de los algoritmos pueden llegar a producir diferentes resultados de las predicciones ya sea por diferentes factores, pueden ser por un nombre del comando y por ello se tienenq ue veriicar los datos.
- Esto se divide entre el:
    - Entrenamiento que es el que genera el modelo
    - Test este se encarga de evaluar la vondad del modelo.
- A su vez nos podemos encontrar con diferentes problemas que se peuden presentar en el aprendizaje como:
    - Sobreajuste o overfitting que genera que el modelo memorize los datos y no extraiga de forma general el conocimiento.
    - Subajuste o underfitting que ocaciona que no pueda discriminar bien el cunjunto de entrenamiento.
- Estos fenomenos pueden ser ser evaluados por medio de
    - Error de entrenamiento
$$
\text{MAE} = \frac{1}{n} \sum_{i=1}^{n} \left| y_i - \hat{y}_i \right|
$$
MAE (Mean Absolute Error / Error Absoluto Medio): mide el promedio de las diferencias absolutas entre valores reales y valores predichos.

𝑛: número de observaciones.

𝑦𝑖: valor real.

𝑦^𝑖: valor predicho.

    - Error de generalización
$$
\text{Exactitud} = \frac{100}{n} \sum_{i=1}^{n} \mathbf{1}(y_i = \hat{y}_i)
$$
Exactitud mide el porcentaje de predicciones correctas.

𝑛 = número total de observaciones.
𝑦𝑖 = valor real de la observación 
𝑖.𝑦^𝑖 = valor predicho por el modelo.

1(𝑦𝑖=𝑦^𝑖) = función indicadora que vale 1 si la predicción coincide con el valor real, y 0 en caso contrario.

    

## Validacion de hold-out

- Este el fin que se busca es que la base de datos que se usan se dividan entre el entrenemiento y el test usualmente es de 70% de entrenamiento y 30% de pruebas.
- La entaja es que es sencillo de realizar y economica.
- La desventaja es tiene la limitacion en la escasa representatividad de los resultados.
- Los métodos de división son
    - Aleatoria
    - Secuencial
    - Dirigida
    - Estratificada.

## Validación Cruzada
- Consiste en partir el conjunto original de datos en subconjuntos que s elos conocen como Bolsas o Folds, esto debe ser de igual tamaño.
- Una vez partido el conjunto se deben realizar las validaciones de tipo hold-out, alternando cada fold como test.
- lo que obtenemos es una metrica la cual nos indica el promedio de los resultados.
- En el caso especial de leave-one-out lo que pase es que sustituye el promedio por concatenación.


## Ajuste de Parámetros y Validación Anidada
- Los algoritmos de aprendizaje supervisado pueden poseer uno o varios parámetros de configuración logrando que regulen su funcionamiento y permiten adaptar el algoritmo.
- Se requiere un proceso de selección cuidadosa de parámetros.
- La validación anidada combina:
    - Conjunto de entrenamiento.
    - Conjunto de validación que ayuda para ajustar los parámetros.
    - Conjunto de test que sirve para evaluación final.


## Evaluación en Regresión
- Se evaluan por medio de la metricas absolutas
- MAE o Mean Absolute Error.
$$
\text{MAE} = \frac{1}{n} \sum_{i=1}^{n} \left| y_i - \hat{y}_i \right|
$$
    - RMSE o Root Mean Squared Error.
$$
\text{RMSE} = \sqrt{ \frac{1}{n} \sum_{i=1}^{n} \left( y_i - \hat{y}_i \right)^2 }
$$

- Y con metricas relativas
    - MAPE o Mean Absolute Percentage Error que mide el error relativo respecto al valor real.
$$
\text{MAPE} = \frac{100}{n} \sum_{i=1}^{n} \left| \frac{y_i - \hat{y}_i}{y_i} \right|
$$


## Evaluación en clasificación
- En esta se encarga de verificar si los datos que se predijeron coinciden con los reales
- La forma que se interpreta es mediante la matriz de confusion que puede ser multiclase o binaria.
- esto se puede medir mediante:
    - Exactitud o Accuracy: porcentaje de predicciones correctas sobre el total.
$$
\text{Exactitud}(C_i) = 100 \cdot \frac{P_u}{\sum_{i=1}^{n} P_u}
$$
    - Sensibilidad o Recall:  capacidad de detectar correctamente los positivos reales.
$$
\text{Sensibilidad} = \frac{TP}{TP + FN}
$$
    - Precisión o Precision: proporción de positivos predichos que realmente son positivos.
$$
\text{Precisión} = 100 \cdot \frac{TP}{TP + FP}
$$
    - F1-Score que hace una media armónica entre sensibilidad y precisión, útil cuando hay desbalance de clases.
$$
F1 = 100 \cdot \frac{2 \cdot \text{Sensibilidad} \cdot \text{Precisión}}{\text{Sensibilidad} + \text{Precisión}}
$$
